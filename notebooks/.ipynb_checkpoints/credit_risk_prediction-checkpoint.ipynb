<h1>Loan/Default</h1>

```python
# all imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split as tts
from sklearn.preprocessing import OneHotEncoder as oneHot, OrdinalEncoder as ordinal
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# used 3 different models for the same data
from sklearn.linear_model import LogisticRegression 
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, classification_report


"""chaning the order of execution of cells will cause erros, execute cells in order"""

print("all imports done")
```

```python
# reading the data

my_df = pd.read_csv("../dataset/german_credit_data.csv")
my_df.head()
```

```python
my_df.info()
```

```python
# dropping unnamed column
# added check so that running it multiple times doesn't throw error
if "Unnamed: 0" in my_df.columns:
    my_df = my_df.drop("Unnamed: 0",axis=1)  # axis = 0 (rows), axis = 1(columns)
    print("Un-named column dropped")


# renaming column names for convenience
my_df.columns = my_df.columns.str.replace(" ","_")
my_df.columns = my_df.columns.str.lower()
my_df.rename(columns={"saving_accounts":"savings_account"}, inplace=True)

my_df.columns
```

```python
# filling NaN values as none
my_df.checking_account = my_df.checking_account.fillna("none")
my_df.savings_account = my_df.savings_account.fillna("none")

my_df.head()
```

```python
# create a new column based on risk, if risk is bad, then the column shows true else flase
# my_df.default = my_df.risk == "bad"


# this created a column of True/False values
# now if we convert that it interger "bad" -> True -> (becomes) 1 and "good" -> False -> (becomes) 0
my_df["default"] = (my_df.risk == "bad").astype(int)
""" [NOTE: my_df. = dot notation does not add column to the dataframe, it creates an attribue, my_df["col_name"] properly adds one column to the df] """

# we can keep the check as "good", then 1 then it becomes 1 for good and 0  for bad
# but as a convention we choose to check for "bad" and keep it as 1 as we are predicting the chance of default
# so of we get a probablity of 0.85, it will mean that chance of deafault is 85%

my_df.columns
```

```python
# now we drop the risk column
if "risk" in my_df.columns:
    my_df = my_df.drop("risk", axis=1)
    print("Risk column dropped")

my_df.columns
```

```python
# checking for missing data
print(my_df.isnull().sum())

my_df['default'].value_counts()

# as we can see, we don't have any null values in our dataset
```

<h2>ML</h2>

```python
# now to split the data

X = my_df.drop("default", axis=1) # drop the column as it will be the y value
y = my_df["default"]

X_train, X_test, y_train, y_test = tts(X, y, random_state=32, stratify=y)
```

```python
my_df.head()
```

```python
# now ordinal/one hot enconding the categorical columns

# the nominal columns, to use one-hot encoder
nominal_cols = ["sex","housing","purpose"]

# the ordinal columns, to use ordinal transformer
ordinal_cols = ["savings_account","checking_account"]

# to set their order, we used .unique to get all these values
savings_order = ["none","little","moderate","quite rich","rich"]
checking_order = ["none","little","moderate","rich"]

# we get the above data by checking the unique values of both the columns and giving them an order
# print("for savings: ",my_df["savings_account"].unique())
# print("for checking: ",my_df["checking_account"].unique())


# preprocessing steps
my_preprocessor = ColumnTransformer(transformers=[
    # for nominal column, use one-hot encoder
    ("nominal", oneHot(handle_unknown="ignore"), nominal_cols),

    # for ordinal columns, use ordinal encoder
    ("ordinal", ordinal(categories=[savings_order,checking_order]), ordinal_cols),
], remainder="passthrough")
print("Preprocessing Completed")
```

```python
# now to check if all is right till here, fit the data
# X_train_transformed = my_preprocessor.fit_transform(X_train)
# X_train_transformed.shape
```

```python
# making pipeline for logistic_regressor
log_reg_pipeline = Pipeline([
    ("preprocessor", my_preprocessor),
    ("classifier", LogisticRegressor(max_iter=10000))
])

# fir on training data and predicting
my_pipeline.fit(X_train, y_train)
log_reg_pred = my_pipeline.predict(X_test)
print("LogisticRegressor put to work")
```

```python
# making pipeline for randomforst
rf_pipeline = Pipeline([
    ("preprocessor", my_preprocessor),
    ("classifier", RandomForestRegressor(random_state=1))
])

# fir on training data and predicting
rf_pipeline.fit(X_train, y_train)
rf_pred = my_pipeline.predict(X_test)
print("RandomForest counting trees")
```

```python
# time to predict!!




print("Accuracy:\t", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
